```python
def plot3d(xpt, ypt, zpt, alpha=0.3, s=2, ax=None):

    fig = None
    if ax is None:
        fig = plt.figure()
        ax = fig.add_subplot(projection='3d')
    
    # Plot points
    ax.scatter(xpt, ypt, zpt, alpha=alpha, s=s)
    
    # Labels
    ax.set_xlabel('X Axis')
    ax.set_ylabel('Y Axis')
    ax.set_zlabel('Z Axis')

    return fig, ax
```
