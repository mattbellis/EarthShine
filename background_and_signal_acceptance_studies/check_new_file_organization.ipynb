```python
cat = eio.read_catalog("data")
#cat[["dm_model","mDM_min","mDM_max","depth_min","depth_max","disk_radius","n_rows"]]

plot_params = []
#                     model   , mDM_min, mDM_max, d_min, d_max, r,    emin,  mass
plot_params.append(['floating', 100000,   900000, -4000,   -8,  4000, 100,  100000])
plot_params.append(['floating', 10000,   90000, -4000,   -8,  4000, 100,  10000])
plot_params.append(['floating', 1000,   9000, -4000,   -8,  4000, 100,  1000])
plot_params.append(['floating', 100,   900, -1000,   -8,  1000, 10,  100])
plot_params.append(['core', 100000,   900000, -4000,   -8,  40, 100,  100000])
plot_params.append(['core', 10000,   90000, -4000,   -8,  40, 100,  10000])
plot_params.append(['core', 1000,   9000, -4000,   -8,  40, 100,  1000])
plot_params.append(['core', 100,   900, -1000,   -8,  40, 10,  100])

#inner_detector = True
inner_detector = False

for plot_param in plot_params:
    model = plot_param[0]
    
    mDM_min = plot_param[1]
    mDM_max = plot_param[2]
    depth_min = plot_param[3]
    depth_max = plot_param[4]
    disk_radius = plot_param[5]
    
    emin = plot_param[6]
    mass = plot_param[7]
    
    df, params = eio.load_many(cat, masses=[mass], dm_model=model, stage="combined",
                          depth_min=depth_min, depth_max=depth_max, disk_radius=disk_radius, \
                         mDM_min=mDM_min, mDM_max=mDM_max, eloss='ave')   # add filters until exactly 1 matches
    
    dftmp,figs = diag.e_and_pt_plots(df, params, min_efinal=emin, inner_detector=inner_detector, plotdir="plots")
    
    dftmp,figs = diag.kinematic_diagnostic(df, params, min_efinal=emin, inner_detector=inner_detector, plotdir="plots")
    
    dftmp,figs = diag.origin_plots(df,  params, min_efinal=emin, inner_detector=inner_detector, plotdir="plots")
```
